```python
import pandas as pd
import re

# Create a sample DataFrame if 'reviews.csv' doesn't exist
sample_data = {
    'Review_Text': ['Great product!', 'Not good.', None, 'Amazing!', 'Terrible.'],
    'Star_Rating': [5, 2, 4, 5, 1]  # Assuming you have this column for later steps
}
df = pd.DataFrame(sample_data)
df.to_csv('reviews.csv', index=False)  # Save it for future use
```

```python
from textblob import TextBlob
import pandas as pd
import os

# Step 1: Ensure the DataFrame is loaded (from previous cleaning step)
# If 'df' is not defined, load from the cleaned CSV
if 'df' not in locals():
    try:
        df = pd.read_csv('cleaned_reviews.csv')
        print("Loaded DataFrame from 'cleaned_reviews.csv'.")
    except FileNotFoundError:
        print("Error: 'cleaned_reviews.csv' not found. Please run the cleaning script first or check the file path.")
        exit()

# Step 2: Check if the required column exists
if 'Review_Text' not in df.columns:
    print("Error: 'Review_Text' column not found in the DataFrame. Please ensure the cleaning script ran successfully.")
    print("Available columns:", list(df.columns))
    exit()

# Define a function to get the sentiment polarity score
def get_sentiment_polarity(text):
    """
    Takes a text string and returns the sentiment polarity score using TextBlob.
    Polarity ranges from -1 (negative) to 1 (positive).
    """
    if pd.isna(text):  # Handle NaN values if any
        return None
    return TextBlob(text).sentiment.polarity

# Apply the function to the 'Review_Text' column (corrected from 'Cleaned_Reviews') and create a new column 'Sentiment_Score'
df['Sentiment_Score'] = df['Review_Text'].apply(get_sentiment_polarity)

# Optional: Display the first few rows to verify (using 'Review_Text' instead of 'Cleaned_Reviews')
print(df[['Review_Text', 'Sentiment_Score']].head())

# Optional: Save the updated DataFrame
df.to_csv('reviews_with_sentiment.csv', index=False)
print("Updated data saved to 'reviews_with_sentiment.csv'.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'df' is your DataFrame with columns 'Star_Rating' (1-5) and 'Sentiment_Score'
# If not loaded, uncomment and adjust: df = pd.read_csv('reviews_with_sentiment.csv')

# Group by Star_Rating to get count of reviews and average Sentiment_Score
rating_summary = df.groupby('Star_Rating').agg(
    Review_Count=('Star_Rating', 'size'),  # Count of reviews
    Avg_Sentiment=('Sentiment_Score', 'mean')  # Average Sentiment_Score
).reset_index()

# Create a bar chart
plt.figure(figsize=(10, 6))

# Use a colormap to color bars based on Avg_Sentiment (e.g., from red for negative to green for positive)
colors = plt.cm.RdYlGn((rating_summary['Avg_Sentiment'] + 1) / 2)  # Normalize to 0-1 for colormap

# Plot bars with colors
bars = plt.bar(rating_summary['Star_Rating'], rating_summary['Review_Count'], color=colors)

# Add colorbar for reference
sm = plt.cm.ScalarMappable(cmap='RdYlGn', norm=plt.Normalize(vmin=-1, vmax=1))
sm.set_array([])
cbar = plt.colorbar(sm, ax=plt.gca())
cbar.set_label('Average Sentiment Score')

# Add labels and title
plt.xlabel('Star Rating')
plt.ylabel('Count of Reviews')
plt.title('Sentiment vs. Star Rating')
plt.xticks(rating_summary['Star_Rating'])  # Ensure x-ticks are 1-5

# Show the plot
plt.show()
```
